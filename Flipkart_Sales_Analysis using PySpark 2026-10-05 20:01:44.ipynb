Flipkart Sales Analysis using PySpark in Databricks

```python
from pyspark.sql import functions as F
from pyspark.sql.types import *
```

```python
from pyspark.sql import functions as F
import pandas as pd

file_path = "https://github.com/mskalaiwani00294-lang/Flipkart-Sales-Analysis-Using-PySpark-Databricks/raw/refs/heads/main/flipkart_style_dataset_200_rows.csv"
df = spark.createDataFrame(pd.read_csv(file_path))
display(df)
```

```python
#head first 10 rows
df.show(10)
```

```python
print("Total Rows:", df.count())
print("Total Columns:", len(df.columns))
print("Column Names:")
print(df.columns)
df.printSchema()
```

```python
# Check missing values
from pyspark.sql import functions as F

df.select([
    F.count(F.when(F.col(c).isNull(), c)).alias(c)
    for c in df.columns
]).show()
```

```python
# Check duplicate rows
duplicate_count = df.count() - df.dropDuplicates().count()

print("Duplicate Rows:", duplicate_count)
```

```python
#data transfermation
df=df.withColumn(
    "Total_Amount",
                 F.col("Quantity")*F.col("Unit_Price")
)
display(df)
```

**Flipkart Sales Analysis** – **PySpark**

```python
#1. Which Category Has the Highest Sales?
from pyspark.sql import functions as F
display(
    df.groupBy("Product_Category").agg(
        F.sum("Total_Amount").alias("Total_Sales")
    ).orderBy(
        F.col("Total_Sales").desc()
    )
)
```

```python
#2. Which Product Sells the Most?
display(
df.groupBy("Product_Name").agg(
    F.sum("Quantity").alias("Total_Quantity")
).orderBy(
    F.col("Total_Quantity").desc()
).limit(5)
)
```

```python
#3. Which City Has the Highest Sales?

display(
df.groupBy("City").agg(
    F.sum("Total_Amount").alias("Total_Sales")
).orderBy(
    F.col("Total_Sales").desc()
)
)
```

```python
#4. Which Category Has the Highest Sales Amount on Average?
display(
df.groupBy("Product_Category").agg(
    F.avg("Total_Amount").alias("Average_Sales")
).orderBy(
    F.col("Average_Sales").desc()
)
)
```

```python
#5. Top 5 Sales Records
display(
df.orderBy(
    F.col("Total_Amount").desc()
).limit(5)
)
```

```python
#Bonus: Total Sales and Total Quantity
df.select(
    F.sum("Total_Amount").alias("Total_Sales"),
    F.sum("Quantity").alias("Total_Quantity")
).show()
```
